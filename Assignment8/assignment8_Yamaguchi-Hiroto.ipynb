## 課題1

以下の図を再現してください

- 年齢の分布を、異なる階級幅を使って、4つのサブプロットをわけて描画する
   - 階級幅を適当で決めてよい
- 各サブプロットには、
    - x軸ラベル、y軸ラベル、サブプロット番号(a-d)を指定する
    - 水平方向のグリッド線を追加する
    - ヒストグラムの色を`lightblue`に変更する

![](./Figure1.png)

```python
import matplotlib.pyplot as plt
import japanize_matplotlib
import numpy as np


plt.rcParams["xtick.minor.visible"] = True
plt.rcParams["ytick.minor.visible"] = True
plt.rcParams["xtick.direction"] = "in"
plt.rcParams["ytick.direction"] = "in"

age = df["age"].dropna()

fig = plt.figure(figsize=(11,9))


ax1 = plt.subplot(321)
ax1.hist(age,
         bins=40,
         color="lightblue",
         edgecolor="white", 
         rwidth=0.8,
         alpha=0.6)

ax1.set_title("a", loc="left")
ax1.set_xlabel("年齢(歳)")
ax1.set_ylabel("人数")




ax2 = plt.subplot(322)
ax2.hist(age,
         bins=30,
         color="lightblue",
         rwidth=0.8,
         edgecolor="white",
         alpha=0.6)

ax2.set_title("b", loc="left")
ax2.set_xlabel("年齢(歳)")
ax2.set_ylabel("人数")




ax3 = plt.subplot(323)
ax3.hist(age,
         bins=15,
         color="lightblue",
         edgecolor="white",
         rwidth=0.8,
         alpha=0.6)

ax3.set_title("c", loc="left")
ax3.set_xlabel("年齢(歳)")
ax3.set_ylabel("人数")
ax3.set_yticks(range(0, 200, 50))


ax4 = plt.subplot(324)
ax4.hist(age,
         bins=10,
         color="lightblue",
         edgecolor="white",
         rwidth=0.8,
         alpha=0.6)

ax4.set_title("d", loc="left")
ax4.set_xlabel("年齢(歳)")
ax4.set_ylabel("人数")



for ax in [ax1, ax2, ax3, ax4]:
    ax.set_axisbelow(True)
    ax.grid(axis="y", linestyle="-", alpha=0.4)

plt.tight_layout()
plt.show()
```

## 課題2

以下の図を再現してください

- 男性乗客と女性乗客の年齢の分布をそれぞれ描画する
- 全乗客の年齢分布を示す灰色ヒストグラムも描画する
- 図のタイトル、x軸ラベル、y軸ラベル、凡例を設定する

```python
import numpy as np

male = df[df["sex"]=="male"]["age"].dropna()
female = df[df["sex"]=="female"]["age"].dropna()

fig, (ax5, ax6) = plt.subplots(1, 2, figsize=(12, 5))

bins = np.arange(0, 82, 1.6)

ax5.hist(age,
         bins=bins,
         color="gray",
         alpha=0.9,
         label="すべての乗客")


ax5.hist(male,
         bins=bins,
         color="blue",
         alpha=0.7,
         label="男性乗客")

ax5.set_title("男性乗客")
ax5.set_xlabel("乗客の年齢(歳)")
ax5.set_ylabel("人数")
ax5.legend()





ax6.hist(age,
         bins=bins,
         color="gray",
         alpha=0.9,
         label="すべての乗客")


ax6.hist(female,
         bins=bins,
         color="red",
         alpha=0.7,
         label="女性乗客")

ax6.set_title("女性乗客")
ax6.set_xlabel("乗客の年齢(歳)")
ax6.set_ylabel("人数")
ax6.legend()

for ax in [ax5, ax6]:
    ax.set_axisbelow(True)
    ax.grid(axis="y", linestyle="-", alpha=0.4)

plt.tight_layout()
plt.show()




```
